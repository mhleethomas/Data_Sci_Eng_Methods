```python
import pandas as pd
import math
```

```python
# Store values in a dictionary
values = {
    'y1': 3,
    'y2': 7,
    'y3': 9,
    'y4': 5,
    'y5': 4,
    'y6': 8,
    'lambda': 0.1
}

# Print the values to verify
print(values)
```

# h1

```python
# Define the groups
h1_left_keys = ['y1', 'y2', 'y3']
h1_right_keys = ['y4', 'y5', 'y6']
```

```python
# Create group1 and group2 dictionaries
h1_left = {key: values[key] for key in h1_left_keys}
h1_right = {key: values[key] for key in h1_right_keys}

# Create a DataFrame to display the groups
table_df = pd.DataFrame({
    'h1_left': [f"Y1: {h1_left['y1']:.3f}", f"Y2: {h1_left['y2']:.3f}", f"Y4: {h1_left['y3']:.3f}"],
    'h1_right': [f"Y3: {h1_right['y4']:.3f}", f"Y5: {h1_right['y5']:.3f}", f"Y6: {h1_right['y6']:.3f}"]
})

# Reset the index of the DataFrame to avoid showing the first column
table_df.reset_index(drop=True, inplace=True)

# Print the table to verify
print(table_df)
```

```python
# Calculate the average for each group
avg_h1_left = sum(h1_left.values()) / len(h1_left)
avg_h1_right = sum(h1_right.values()) / len(h1_right)

# Print the averages to verify
print(f"Average of h2_left: {avg_h1_left:.2f}")
print(f"Average of h2_right: {avg_h1_right:.2f}")
```

```python
# Calculate residuals
h1_residuals = {key: values[key] - (avg_h1_left if key in h1_left_keys else avg_h1_right) * values['lambda'] for key in values if key != 'lambda'}

# Store the residuals with denoted names for further calculation
h1_residuals = {f"h1_r{i}": value for i, (key, value) in enumerate(h1_residuals.items(), start=1)}

# Print the stored residuals to verify
print("Residuals:")
for key, value in h1_residuals.items():
    print(f"{key}: {value:.2f}")
```

# h2

```python
# Define the new groups using residuals
h2_left_keys = ['h1_r1', 'h1_r3', 'h1_r5']
h2_right_keys = ['h1_r2', 'h1_r4', 'h1_r6']
```

```python
# Create group1 and group2 dictionaries using residuals
h2_left = {key: h1_residuals[key] for key in h2_left_keys}
h2_right = {key: h1_residuals[key] for key in h2_right_keys}

# Create a DataFrame to display the groups
table_df2 = pd.DataFrame({
    'h2_left': [f"{key}: {h2_left[key]:.2f}" for key in h2_left_keys],
    'h2_right': [f"{key}: {h2_right[key]:.2f}" for key in h2_right_keys]
})

# Reset the index of the DataFrame to avoid showing the first column
table_df2.reset_index(drop=True, inplace=True)

# Print the table to verify
print(table_df2)
```

```python
# Calculate the average for each group
avg_h2_left = sum(h2_left.values()) / len(h2_left)
avg_h2_right = sum(h2_right.values()) / len(h2_right)

# Print the averages to verify
print(f"Average of h2_left: {avg_h2_left:.2f}")
print(f"Average of h2_right: {avg_h2_right:.2f}")
```

```python
# Calculate residuals
h2_residuals = {key: h1_residuals[key] - (avg_h2_left if key in h2_left_keys else avg_h2_right) * values['lambda'] for key in h1_residuals}

# Store the residuals with denoted names for further calculation
h2_residuals = {f"h2_r{i}": value for i, (key, value) in enumerate(h2_residuals.items(), start=1)}


# Print residuals with denoted names
print("Residuals:")
for i, (key, value) in enumerate(h2_residuals.items(), start=1):
    print(f"{key}: {value:.2f}")
```

# h3

```python
# Define the new groups using residuals
h3_left_keys = ['h2_r1', 'h2_r2', 'h2_r3']
h3_right_keys = ['h2_r4', 'h2_r5', 'h2_r6']
```

```python
# Create group1 and group2 dictionaries using residuals
h3_left = {key: h2_residuals[key] for key in h3_left_keys}
h3_right = {key: h2_residuals[key] for key in h3_right_keys}

# Create a DataFrame to display the groups
table_df3 = pd.DataFrame({
    'h3_left': [f"{key}: {h3_left[key]:.2f}" for key in h3_left_keys],
    'h3_right': [f"{key}: {h3_right[key]:.2f}" for key in h3_right_keys]
})

# Reset the index of the DataFrame to avoid showing the first column
table_df3.reset_index(drop=True, inplace=True)

# Print the table to verify
print(table_df3)
```

```python
# Calculate the average for each group
avg_h3_left = sum(h3_left.values()) / len(h3_left)
avg_h3_right = sum(h3_right.values()) / len(h3_right)

# Print the averages to verify
print(f"Average of h3_left: {avg_h3_left:.2f}")
print(f"Average of h3_right: {avg_h3_right:.2f}")
```

```python
# Calculate residuals
h3_residuals = {key: h2_residuals[key] - (avg_h3_left if key in h3_left_keys else avg_h3_right) * values['lambda'] for key in h2_residuals}

# Store the residuals with denoted names for further calculation
h3_residuals = {f"h3_r{i}": value for i, (key, value) in enumerate(h3_residuals.items(), start=1)}

# Print residuals with denoted names
print("Residuals:")
for i, (key, value) in enumerate(h3_residuals.items(), start=1):
    print(f"{key}: {value:.2f}")
```

```python
# Define the groups for each stage
h1_groups = {'left': h1_left_keys, 'right': h1_right_keys}
h2_groups = {'left': h2_left_keys, 'right': h2_right_keys}
h3_groups = {'left': h3_left_keys, 'right': h3_right_keys}

# Function to determine the group average for a given key
def get_group_average(key, stage_groups, stage_averages):
    for group, keys in stage_groups.items():
        if key in keys:
            return stage_averages[group]
    return None

# Generate the key mapping programmatically
key_mapping = {f'y{i+1}': (f'h1_r{i+1}', f'h2_r{i+1}', f'h3_r{i+1}') for i in range(len(values) - 1)}

# Calculate the h_boost for each key
h_boost = {}
for key in values:
    if key == 'lambda':
        continue
    
    h1_key, h2_key, h3_key = key_mapping[key]
    avg_h1 = get_group_average(key, h1_groups, {'left': avg_h1_left, 'right': avg_h1_right})
    avg_h2 = get_group_average(h1_key, h2_groups, {'left': avg_h2_left, 'right': avg_h2_right})
    avg_h3 = get_group_average(h2_key, h3_groups, {'left': avg_h3_left, 'right': avg_h3_right})
    
    if avg_h1 is None or avg_h2 is None or avg_h3 is None:
        raise ValueError(f"Average not found for key: {key}")
    
    h_boost[key] = values['lambda'] * (avg_h1 + avg_h2 + avg_h3)

# Print the h_boost values
print("HBoost values:")
for key, value in h_boost.items():
    print(f"hboost({key.replace('y', 'x')}): {value:.2f}")
```

```python
x = 0.3237 # Value to calculate the sigmoid
```

```python
# Sigmoid function
def sigmoid(x):
    return 1 / (1 + math.exp(-x))

sigmoid_value = sigmoid(x)

print(f"Sigmoid({x}) = {sigmoid_value:.3f}")
```

```python
# Derivative of the sigmoid function
def dsigmoid(x):
    sig = sigmoid(x)
    return sig * (1 - sig)

# Calculate the derivative at x
dsigmoid_value = dsigmoid(x)

print(f"Derivative of Sigmoid({x}) = {dsigmoid_value:.3f}")
```
